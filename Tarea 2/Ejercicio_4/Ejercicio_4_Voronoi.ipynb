# Ejercicio 4. Diagramas de Voronoi

Se representa una ciudad pequeña mediante un plano cartesiano y se ubican droguerías, centros de atención de salud y colegios. Para cada tipo de servicio se construye un diagrama de Voronoi.

## Interpretación

Cada región representa la zona cuyo punto más cercano corresponde al establecimiento asociado. Una región extensa puede indicar una zona con menor cobertura espacial, aunque el diagrama por sí solo no demuestra que falte un establecimiento.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import Voronoi, voronoi_plot_2d

droguerias = np.array([
    [2, 8], [7, 7], [5, 2], [9, 3], [1, 4]
])

salud = np.array([
    [3, 6], [8, 8], [7, 2], [2, 2]
])

colegios = np.array([
    [1, 9], [5, 8], [8, 5], [4, 3], [9, 1]
])

def graficar_voronoi(puntos, titulo):
    vor = Voronoi(puntos)

    fig = voronoi_plot_2d(
        vor,
        show_vertices=False,
        line_colors="black",
        line_width=1
    )

    plt.scatter(puntos[:, 0], puntos[:, 1], s=80, zorder=3)

    for i, p in enumerate(puntos):
        plt.text(p[0] + 0.1, p[1] + 0.1, str(i + 1))

    plt.xlim(0, 10)
    plt.ylim(0, 10)
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.title(titulo)
    plt.grid()
    plt.show()

graficar_voronoi(droguerias, "Diagrama de Voronoi - Droguerías")
graficar_voronoi(salud, "Diagrama de Voronoi - Centros de atención de salud")
graficar_voronoi(colegios, "Diagrama de Voronoi - Colegios")
```

```python
# Comparación visual de las ubicaciones
plt.figure(figsize=(8, 8))
plt.scatter(droguerias[:,0], droguerias[:,1], marker="o", label="Droguerías")
plt.scatter(salud[:,0], salud[:,1], marker="s", label="Salud")
plt.scatter(colegios[:,0], colegios[:,1], marker="^", label="Colegios")
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Distribución de servicios en la ciudad")
plt.legend()
plt.grid()
plt.show()
```
